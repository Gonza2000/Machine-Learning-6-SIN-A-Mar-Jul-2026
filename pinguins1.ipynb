```python
# Import core data science and visualization libraries
import numpy as np          # NumPy is used for numerical operations and array manipulations
import pandas as pd         # Pandas is used for data analysis and structure manipulation (DataFrames)
import matplotlib.pyplot as plt # Matplotlib's pyplot is used for creating basic plots and figures
import seaborn as sns       # Seaborn is used for advanced, attractive statistical visualizations

# Set global plotting styling parameters
plt.rcParams['figure.figsize'] = (14, 7)  # Sets default figure dimensions to 14 inches wide by 7 inches high
plt.style.use('fivethirtyeight')         # Applies the 'fivethirtyeight' styling template to all plots

# Import notebook display utilities
from IPython.display import display      # Allows displaying multiple objects, DataFrames, and rich content in Jupyter cells
```

```python
# Load the penguins dataset from the local CSV file uploaded to the repository
df = pd.read_csv('penguins_size.csv')

# Display the first 5 rows of the DataFrame to inspect the data structure
df.head()
```

```python
# Get the dimensions of the DataFrame (returns a tuple representing: (rows, columns))
df.shape
```

```python
# Count the number of missing (null) values in each column
df.isnull().sum()
```

```python
# Calculate the maximum percentage of null (missing) values across all columns
# df.isnull().sum() counts nulls per column
# Dividing by len(df) and multiplying by 100 gets the percentage for each column
# .max() retrieves the highest percentage value among all columns
(df.isnull().sum() / len(df) * 100).max()
```

```python
# Drop all rows that contain missing (null) values in the DataFrame
df = df.dropna()

# Check the unique values in the 'sex' column after dropping nulls
df['sex'].unique()
```

```python
# Filter out rows where the 'sex' column contains '.' (invalid data)
df = df[df['sex'] != '.']

# Check the unique values in the 'sex' column to confirm cleaning
df['sex'].unique()
```

# Markdown Visualization

```python
# Create a scatter plot to visualize the relation between culmen length and depth
# Color the data points based on the penguin species using the 'Dark2' color palette
sns.scatterplot(x='culmen_length_mm', y='culmen_depth_mm', data=df, hue='species', palette='Dark2')
```

```python
# Create a pair plot to visualize pairwise relationships between all numeric variables
# Color the data points by species using the 'Dark2' color palette
sns.pairplot(df, hue='species', palette='Dark2')
```

El segundo gráfico (pairplot) refleja las relaciones cruzadas y distribuciones individuales de todas las variables numéricas del dataset (`culmen_length_mm`, `culmen_depth_mm`, `flipper_length_mm`, `body_mass_g`), segmentadas por especie (`species`). El tercer gráfico (boxplot a través de catplot) refleja la distribución de la longitud del culmen (`culmen_length_mm`) para cada una de las tres especies de pingüinos (*Adelie*, *Chinstrap* y *Gentoo*).
```python
# Convert categorical variables in the DataFrame into dummy/indicator variables (one-hot encoding)
pd.get_dummies(df)
```

```python
# Drop the target variable 'species' from the features
# Perform one-hot encoding on the remaining categorical features
# Use drop_first=True to omit the first category of each variable to prevent multicollinearity (dummy variable trap)
pd.get_dummies(df.drop('species', axis=1), drop_first=True)
```

# ¿Qué hace?

La función `pd.get_dummies` realiza la codificación **one-hot (one-hot encoding)** sobre las variables categóricas del DataFrame (como `island` y `sex`). Convierte cada categoría en una columna binaria con valores `1` o `0`.

En el segundo caso:
- `df.drop('species', axis=1)`: Elimina la columna objetivo `species` para quedarse únicamente con las variables de entrada (características).
- `drop_first=True`: Elimina la primera columna de categoría de cada variable (por ejemplo, para `sex` solo queda `sex_MALE`). Esto evita el problema de la **multicolinealidad** (la trampa de la variable ficticia o *dummy variable trap*) al entrenar modelos de aprendizaje automático.

```python
# Split the dataset into features (X) and target (y)
X = pd.get_dummies(df.drop('species', axis=1), drop_first=True)
y = df['species']

# Import train_test_split utility from sklearn
from sklearn.model_selection import train_test_split

# Split the data into training (70%) and testing (30%) sets
# Set random_state to the last 3 digits of the ID card (583)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=583)
```

# Decision Tree Classifier
## Default Hyperparameters

```python
# Import the Decision Tree Classifier from scikit-learn
from sklearn.tree import DecisionTreeClassifier

# Instantiate the DecisionTreeClassifier with default settings
model = DecisionTreeClassifier()

# Fit/train the model using the training data
model.fit(X_train, y_train)

# Make predictions on the test dataset
base_pred = model.predict(X_test)
```

# Evaluation

```python
# Import evaluation metrics from scikit-learn
from sklearn.metrics import confusion_matrix, classification_report

# Compatibility wrapper for plot_confusion_matrix (removed in scikit-learn 1.2+)
try:
    from sklearn.metrics import plot_confusion_matrix
except ImportError:
    from sklearn.metrics import ConfusionMatrixDisplay
    plot_confusion_matrix = ConfusionMatrixDisplay.from_estimator

# Compute the confusion matrix
confusion_matrix(y_test, base_pred)

# Set the matplotlib style to 'classic'
plt.style.use('classic')

# Plot the confusion matrix
plot_confusion_matrix(model, X_test, y_test)
```

```python
# Print the classification report showing metrics like precision, recall, f1-score, and support
print(classification_report(y_test, base_pred))
```

# ¿Qué es Support?

El **support** (soporte) en el reporte de clasificación indica el número de ocurrencias reales de cada clase en el conjunto de prueba (`y_test`). Representa la cantidad de muestras verdaderas de cada especie de pingüino en los datos de prueba, lo cual es útil para evaluar si las clases del conjunto de datos están balanceadas o no.

```python
# Import plot_tree function from sklearn to visualize the decision tree structure
from sklearn.tree import plot_tree

# Set the figure size of the plot
plt.figure(figsize=(12, 8))

# Draw the decision tree structure
plot_tree(model);
```

```python
# Create a figure setting custom dimensions and resolution (DPI)
plt.figure(figsize=(12, 8), dpi=150)

# Plot the decision tree structure with filled nodes and labeled feature names
plot_tree(model, filled=True, feature_names=X.columns);
```
