```python
# Import core data science and visualization libraries
import numpy as np          # NumPy is used for numerical operations and array manipulations
import pandas as pd         # Pandas is used for data analysis and structure manipulation (DataFrames)
import matplotlib.pyplot as plt # Matplotlib's pyplot is used for creating basic plots and figures
import seaborn as sns       # Seaborn is used for advanced, attractive statistical visualizations

# Set global plotting styling parameters
plt.rcParams['figure.figsize'] = (14, 7)  # Sets default figure dimensions to 14 inches wide by 7 inches high
plt.style.use('fivethirtyeight')         # Applies the 'fivethirtyeight' styling template to all plots

# Import notebook display utilities
from IPython.display import display      # Allows displaying multiple objects, DataFrames, and rich content in Jupyter cells
```

```python
# Load the penguins dataset from the local CSV file uploaded to the repository
df = pd.read_csv('penguins_size.csv')

# Display the first 5 rows of the DataFrame to inspect the data structure
df.head()
```

```python
# Get the dimensions of the DataFrame (returns a tuple representing: (rows, columns))
df.shape
```

```python
# Count the number of missing (null) values in each column
df.isnull().sum()
```

```python
# Calculate the maximum percentage of null (missing) values across all columns
# df.isnull().sum() counts nulls per column
# Dividing by len(df) and multiplying by 100 gets the percentage for each column
# .max() retrieves the highest percentage value among all columns
(df.isnull().sum() / len(df) * 100).max()
```

```python
# Drop all rows that contain missing (null) values in the DataFrame
df = df.dropna()

# Check the unique values in the 'sex' column after dropping nulls
df['sex'].unique()
```

```python
# Filter out rows where the 'sex' column contains '.' (invalid data)
df = df[df['sex'] != '.']

# Check the unique values in the 'sex' column to confirm cleaning
df['sex'].unique()
```

# Markdown Visualization

```python
# Create a scatter plot to visualize the relation between culmen length and depth
# Color the data points based on the penguin species using the 'Dark2' color palette
sns.scatterplot(x='culmen_length_mm', y='culmen_depth_mm', data=df, hue='species', palette='Dark2')
```

```python
# Create a pair plot to visualize pairwise relationships between all numeric variables
# Color the data points by species using the 'Dark2' color palette
sns.pairplot(df, hue='species', palette='Dark2')
```
